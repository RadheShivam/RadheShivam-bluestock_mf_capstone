1. **Risk Analysis:** Some schemes exhibited significantly higher VaR and CVaR values, indicating greater downside risk during adverse market conditions.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Load CSV Dataset

```python
nav = pd.read_csv(
    "../data/processed/02_nav_history_cleaned.csv"
)

nav["date"] = pd.to_datetime(nav["date"])

nav = nav.sort_values(
    ["amfi_code", "date"]
)
```

Compute Daily Returns

```python
nav["daily_return"] = (
    nav.groupby("amfi_code")["nav"]
    .pct_change()
)
```

Compute VaR and CVaR

```python
risk_report = []

for fund, group in nav.groupby("amfi_code"):

    returns = (
        group["daily_return"]
        .dropna()
    )

    var_95 = np.percentile(
        returns,
        5
    )

    cvar_95 = (
        returns[
            returns <= var_95
        ]
        .mean()
    )

    risk_report.append({

        "amfi_code": fund,

        "VaR_95": var_95,

        "CVaR_95": cvar_95

    })

risk_report = pd.DataFrame(
    risk_report
)
```

Save CSV File

```python
risk_report.to_csv(

    "../reports/var_cvar_report.csv",

    index=False

)

risk_report.head()
```

2. **Sharpe Ratio Trends:** Rolling 90-day Sharpe Ratios revealed changing risk-adjusted performance over time, with certain funds consistently outperforming their peers.


```python
nav["rolling_sharpe"] = (
    nav.groupby("amfi_code")["daily_return"]
    .transform(
        lambda x: (
            x.rolling(90).mean()
            /
            x.rolling(90).std()
        ) * np.sqrt(252)
    )
)
```

Select 5 Key Funds

```python
top_funds = nav["amfi_code"].unique()[:5]
```

Plot the Sharpe Ratios and png

```python
plt.figure(figsize=(12,6))

for fund in top_funds:

    subset = nav[
        nav["amfi_code"] == fund
    ]

    plt.plot(
        subset["date"],
        subset["rolling_sharpe"],
        label=str(fund)
    )

plt.title("Rolling 90-Day Sharpe Ratio")
plt.xlabel("Date")
plt.ylabel("Sharpe Ratio")

plt.legend()

plt.grid(True)

plt.savefig(
    "../reports/charts/rolling_sharpe_chart.png"
)

plt.show()
```

3. **Investor Cohort Behaviour:** Investors who started investing in earlier cohorts contributed substantially higher total investment amounts compared to recent cohorts.


Load Investor Data

```python
investor = pd.read_csv(
    "../data/raw/08_investor_transactions.csv"
)

investor["transaction_date"] = pd.to_datetime(
    investor["transaction_date"]
)
```

Find First Transaction Year

```python
first_txn = (
    investor.groupby("investor_id")
    ["transaction_date"]
    .min()
    .dt.year
    .reset_index()
)

first_txn.rename(
    columns={
        "transaction_date": "cohort_year"
    },
    inplace=True
)
```

Merge Cohort Information

```python
investor = investor.merge(
    first_txn,
    on="investor_id",
    how="left"
)
```

Cohort Metrics

```python
cohort_summary = (
    investor.groupby("cohort_year")
    .agg(

        avg_sip_amount=(
            "amount_inr",
            "mean"
        ),

        total_invested=(
            "amount_inr",
            "sum"
        )

    )
    .reset_index()
)
```

Top Fund Preference

```python
top_fund = (

    investor.groupby(
        ["cohort_year", "amfi_code"]
    )
    .size()
    .reset_index(
        name="count"
    )

    .sort_values(
        ["cohort_year", "count"],
        ascending=[True, False]
    )

    .drop_duplicates(
        "cohort_year"
    )
)
```

Final Cohort Table

```python
cohort_summary = cohort_summary.merge(

    top_fund[
        [
            "cohort_year",
            "amfi_code"
        ]
    ],

    on="cohort_year"

)

cohort_summary
```

4. **SIP Continuity:** A large proportion of investors were identified as "at-risk," with average SIP gaps exceeding 35 days, suggesting potential discontinuation concerns.


Filter SIP Transactions

```python
sip = investor[
    investor["transaction_type"] == "SIP"
].copy()
```

Sort Transactions

```python
sip = sip.sort_values(
    ["investor_id", "transaction_date"]
)
```

Compute Gap Between Transactions

```python
sip["gap_days"] = (

    sip.groupby("investor_id")
    ["transaction_date"]

    .diff()

    .dt.days
)
```

Keep Investors with 6+ SIPs

```python
eligible = (

    sip.groupby("investor_id")

    .filter(

        lambda x: len(x) >= 6
    )
)
```

Calculate Average Gap

```python
continuity = (

    eligible.groupby("investor_id")

    .agg(

        avg_gap=("gap_days", "mean"),

        sip_count=("transaction_type", "count")
    )

    .reset_index()
)
```

Flag At-Risk Investors

```python
continuity["status"] = np.where(

    continuity["avg_gap"] > 35,

    "at-risk",

    "regular"
)
```

5. **Portfolio Diversification:** HHI analysis showed that most equity funds maintained diversified portfolios, while a smaller subset displayed moderate concentration levels.

Load holdings data

```python
holdings = pd.read_csv("../data/raw/09_portfolio_holdings.csv")

holdings.head()
```

Normalize the weights

```python
holdings["weight"] = holdings.groupby("amfi_code")[
    "weight_pct"
].transform(lambda x: x / x.sum())

holdings.head()
```

Calculate HHI

```python
hhi = (
    holdings.groupby("amfi_code")["weight"]
    .apply(lambda x: (x ** 2).sum())
    .reset_index(name="HHI")
)

hhi.head()
```

Classify the concentration level

```python
def classify_hhi(hhi):
    if hhi < 0.15:
        return "Diversified"
    elif hhi < 0.25:
        return "Moderately Concentrated"
    else:
        return "Highly Concentrated"


hhi["concentration"] = hhi["HHI"].apply(classify_hhi)

hhi.head()
```
